```python
import pandas as pd
```

```python
df_gap_minder = pd.read_csv('../datasets/gapminder_full.csv')
df_health_expenditure = pd.read_csv('../datasets/health_expenditure.csv')
df_literancy_rate = pd.read_csv('../datasets/literancy_rate.csv')
df_mortality_rate = pd.read_csv('../datasets/mortality_rate.csv')
df_gpd = pd.read_csv('../datasets/gpd.csv')
df_drinking_water = pd.read_csv('../datasets/drinking_water.csv')
```

```python
#remover coluna indicator_code de todos os dataframes para manter apenas as colunas necessárias para a aná

def remove_column(df,  list_columns):
    for column_name in list_columns:
        if column_name in df.columns:
            df = df.drop(columns=[column_name])
    return df 

#padronização do country de todos os dataframes removendo espaços e deixando apenas a primeira letra maiúscula para facilitar o merge dos dataframes
def standardize_column(df, column_name):
    df[column_name] = df[column_name].str.strip().str.title()
    return df 

#transformar os dataframes de formato largo para formato longo usando a função melt 
def transform_df_melt(df, list_columns, long_column, value_name):
    df = df.melt(
        id_vars=list_columns,
        var_name=long_column,
        value_name=value_name
    )
    return df
```

```python
#chama o a função remove_column para remover a coluna indicator_code de todos os dataframes

df_health_expenditure = remove_column(df_health_expenditure, ['indicator_name','indicator_code', 'country_code', 'Unnamed: 70'])
df_gpd = remove_column(df_gpd, ['indicator_name','indicator_code', 'country_code', 'Unnamed: 70'])  
df_drinking_water = remove_column(df_drinking_water, ['indicator_name','indicator_code', 'country_code', 'Unnamed: 70'])
df_literancy_rate = remove_column(df_literancy_rate, ['indicator_name','indicator_code', 'country_code', 'Unnamed: 70'])
df_mortality_rate = remove_column(df_mortality_rate, ['indicator_name','indicator_code', 'country_code', 'Unnamed: 70'])
```

```python
#chama o a função standardize_column para padronizar a coluna country de todos os dataframes
column_name = 'country'

df_gap_minder = standardize_column(df_gap_minder, column_name)
df_health_expenditure = standardize_column(df_health_expenditure, column_name)
df_gpd = standardize_column(df_gpd, column_name)
df_drinking_water = standardize_column(df_drinking_water, column_name)
df_literancy_rate = standardize_column(df_literancy_rate, column_name)
df_mortality_rate = standardize_column(df_mortality_rate, column_name)
```

```python
#chama a função transform_df_melt para transformar os dataframes de formato largo para formato longo
list_columns=["country"]
long_column = 'year'

value_name = 'health_expenditure'
df_health_expenditure = transform_df_melt(df_health_expenditure, list_columns, long_column, value_name)

value_name = 'gpd'
df_gpd = transform_df_melt(df_gpd, list_columns, long_column, value_name)

value_name = 'drinking_water'
df_drinking_water = transform_df_melt(df_drinking_water, list_columns, long_column, value_name)

value_name = 'literancy_rate'
df_literancy_rate = transform_df_melt(df_literancy_rate, list_columns, long_column, value_name)

value_name = 'mortality_rate'
df_mortality_rate = transform_df_melt(df_mortality_rate, list_columns, long_column, value_name)

```

```python
df_health_expenditure['year'] = df_health_expenditure['year'].astype(int)

df_gpd['year'] = df_gpd['year'].astype(int)

df_drinking_water['year'] = df_drinking_water['year'].astype(int)

df_literancy_rate['year'] = df_literancy_rate['year'].astype(int)

df_mortality_rate['year'] = df_mortality_rate['year'].astype(int)
```

```python
#df_health_expenditure.head(3)
#df_literancy_rate.head(3)
df_mortality_rate.head(3)
```

```python
df_gap_minder = df_gap_minder.merge(df_health_expenditure, on=['country', 'year'], how='left')
df_gap_minder = df_gap_minder.merge(df_gpd, on=['country', 'year'], how='left')
df_gap_minder = df_gap_minder.merge(df_drinking_water, on=['country', 'year'], how='left')
df_gap_minder = df_gap_minder.merge(df_literancy_rate, on=['country', 'year'], how='left')
df_gap_minder = df_gap_minder.merge(df_mortality_rate, on=['country', 'year'], how='left')

#dividino o gpd por 1 bilhão para facilitar a visualização dos dados
df_gap_minder['gpd_trilhoes'] = (df_gap_minder['gpd'] / 1_000_000_000).round(2)
```
